```python
import pandas as pd
import json

data_path = 'results/pred.json'
data = json.load(open(data_path, 'r'))

tasks = [
    "narrativeqa", "qasper", "multifieldqa_en", "hotpotqa", "2wikimqa", "musique", "gov_report", "qmsum",
    "multi_news", "trec", "triviaqa", "samsum", "passage_count", "passage_retrieval_en", "lcc", "repobench-p"
]
```

```python
model_to_configs = {
    'Meta-Llama-3.1-8B-Instruct': [
        'full',
        '64-16-64'
    ],
    'Mistral-7B-Instruct-v0.2': [
        'full',
        '64-16-64'
    ]
}
```

```python
for model, configs in model_to_configs.items():
    model_data = data[model]

    dataframe = pd.DataFrame(columns=configs)

    for task in tasks:
        row = []
        for c in configs:
            if c == 'full':
                key = 'no_flexicache'
            else:
                parts = c.split('-')
                key = f'flexicache-{parts[0]}-unstable-{parts[1]}-rerank-{parts[2]}-topK'
            row.append(model_data[task][key])
        dataframe = pd.concat(
            [dataframe, pd.DataFrame([row], columns=configs)],
            ignore_index=True
        )

    dataframe.to_csv(f'results/{model}.csv', index=False)
```
