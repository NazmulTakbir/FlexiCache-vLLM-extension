```python
import os, json
import numpy as np
import pandas as pd
from IPython.display import display
import numpy as np
```

```python
models_to_M = {
    "Meta-Llama-3.1-8B-Instruct": 64,
    "Mistral-Small-24B-Instruct-2501": 80,
    "Qwen2.5-32B-Instruct": 128
}
```

```python
# exclude = {'news_summ', 'tv_show_summ', 'patent_summ'}
exclude = {}
K = 128
for model, M in models_to_M.items():
    model_folder = f'Data-Sorted/analysis_outputs/topK-{K}/{model}'
    # model_folder = f'Data/analysis_outputs/{model}'
    rankings = {}
    for task in os.listdir(model_folder):
        if task in exclude:
            continue
        if not os.path.isdir(f'{model_folder}/{task}'):
            continue
        data_file = open(f'{model_folder}/{task}/consensus_M{M}.json', 'r')
        rankings[task] = json.load(data_file)['ranking']
        data_file.close()

    task_names = list(rankings.keys())

    overlap_values = np.array([[0.0 for _ in range(len(task_names))] for _ in range(len(task_names))])

    minimum = None
    for i in range(len(task_names)):
        for j in range(len(task_names)):
            ai = rankings[task_names[i]][:M]
            bj = rankings[task_names[j]][:M]
            overlap_values[i][j] = round(len(set(ai) & set(bj)) / M, 2)

            if minimum is None or overlap_values[i][j] < minimum[1]:
                minimum = ((task_names[i], task_names[j]), overlap_values[i][j])

    df_overlap = pd.DataFrame(overlap_values, index=task_names, columns=task_names)
    print(f'Overlap@{K} Matrix for model: {model} (M={M})')
    display(df_overlap)
    print(f'Minimum Overlap@{K} is between {minimum[0][0]} and {minimum[0][1]}: {minimum[1]}')
    print(f'Mean Overlap@{K}: {round(np.mean(overlap_values), 2)}')
    print(f'\n{'-'*100}\n')
```
